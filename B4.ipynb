```python
def f(x):
    return (x + 3)**2

def gradient(x):
    return 2 * (x + 3)

x = 2
lr = 0.1
iterations = 50

for i in range(iterations):
    x = x - lr * gradient(x)

print("Local Minimum at x:", x)
print("Minimum value:", f(x))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-6, 3, 100)
y = (x + 3)**2

plt.plot(x, y)
plt.scatter(-3, 0, color='red', label='Minimum')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent')
plt.legend()
plt.grid()
plt.show()
```
